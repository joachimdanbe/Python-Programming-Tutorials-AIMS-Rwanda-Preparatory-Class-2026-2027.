One of the major purposes of a variable is to remember a value from one part of a program
so that it can be used in another part of the program.
Here is another example with variables. Before reading on, try to figure out what the values
of x and y will be after the code is executed.

```python
x=3
y=4
z=x+y
z=z+1
x=y
y=5
```

After these four lines of code are executed, x is 4, y is 5 and z is 8. One way to understand
something like this is to take it one line at a time. This is an especially useful technique

for trying to understand more complicated chunks of code. Here is a description of what
happens in the code above:
1. x starts with the value 3 and y starts with the value 4.
2. In line 3, a variable z is created to equal x+y, which is 7.
3. Then the value of z is changed to equal one more than it currently equals, changing it
from 7 to 8.
4. Next, x is changed to the current value of y, which is 4.
5. Finally, y is changed to 5. Note that this does not affect x.
6. So at the end, x is 4, y is 5, and z is 8.

### 3 Arithmetic Operations
In Python, arithmetic operations are used to perform basic mathematical calculations. Python
supports the standard arithmetic operators that you’d find in most programming languages.
These operators include:

| Operator | Description |
| :---: | :--- |
| `+` | Addition |
| `-` | Subtraction |
| `*` | Multiplication |
| `/` | Division |
| `//` | Floor Division |
| `%` | Modulus |
| `**` | Exponentiation |


Here are some examples of how to use these arithmetic operators in Python:

```python
# Addition
result_addition = 5 + 3 # Result: 8
# Subtraction
result_subtraction = 10- 4 # Result: 6
# Multiplication
result_multiplication = 2 * 5 # Result: 10
# Division
result_division = 15 / 3 # Result: 5.0 (always a float)
# Floor Division
result_floor_division = 17 // 4 # Result: 4 (rounded down to nearest integer)
# Modulus
result_modulus = 17 % 4 # Result: 1 (remainder of 17 divided by 4)
# Exponentiation
result_exponentiation = 2 ** 3 # Result: 8 (2 raised to the power of 3)
result = 2 + 3 * 4 # Result: 14 (3 * 4 is done first, then 2 is added)
result_with_parentheses = (2 + 3) * 4 # Result: 20 (2 + 3 is done first, then the result is multiplied by 4)
```

### 4 Logical Operators

| :---: | :--- |

| `or` | Returns True if at least one of the operands is True. |

You can use these logical operators to construct complex conditions and make decisions

```python
# Logical AND
result_and = True and False # Result: False
# Logical OR
result_or = True or False # Result: True
# Logical NOT
result_not = not True # Result: False
```

### 8 Conditional Statements
Conditional statements in Python are used to control the flow of a program based on certain
conditions. These statements allow you to execute specific blocks of code only if a certain
condition is met. The most commonly used conditional statements in Python are:
* if statement: The “if” statement allows you to execute a block of code if a given
condition is true. Here is an example:

```python
temp = eval(input('Enter a temperature in Celsius: '))
f_temp = 9/5 * temp + 32
print('In Fahrenheit, that is', f_temp)

if f_temp > 212:
  print('That temperature is above the boiling point.')
```

* if-else statement: The “if-else” statement allows you to execute one block of code
if a condition is true, and another block of code if the condition is false.

```python
temp = eval(input('Enter a temperature in Celsius: '))
f_temp = 9/5*temp+32
print('In Fahrenheit, that is', f_temp)
if f_temp > 212:
  print('That temperature is above the boiling point.')
else:
  print('That temperature is below the freezing point.')
```

if-elif-else statement: The ”if-elif-else” statement allows you to check multiple
conditions one by one and execute the corresponding block of code for the first condition
that evaluates to true. If none of the conditions are true, the “else” block is executed.

```python
temp = eval(input('Enter a temperature in Celsius: '))
f_temp = 9/5*temp+32
print('In Fahrenheit, that is', f_temp)
if f_temp > 212:
  print('That temperature is above the boiling point.')
elif f_temp < 32:
  print('That temperature is below the freezing point.')
else:
  print('That temperature is between the freezing and boiling point.')
```

### 9 FOR Loop and WHILE loop

A loop as a way to repeat actions like this Fibonnacci sequence.

* a for loop is used when we already know exactly how many times we want to repeat the actions

* a while loop is used when we want to repeat something until a specific condition changes.

```python
f0 = 8000
f1 = 120000

new_f0 = f0
new_f1 = f1
for i in range(5):
  temp = new_f0
  new_f0 = new_f1
  new_f1 = temp + new_f1
print(new_f1)
```

```python
f0 = 8000
f1 = 120000

new_f0 = f0
new_f1 = f1
for i in range(5):
  new_f0, new_f1 = new_f1, new_f1 + new_f0
print(new_f1)
```

range(start, stop, step) will return start, start + 1, ...., stop - 1 if step = 1

```python
number = 1
iteration = 0
while number < 100:
  number += 2
  iteration += 1

print('number = ', number, ' after ', iteration, ' iterations')
```
