# 5. Worked Examples

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner # Class attributes
        self.balance = balance

    def __str__(self):
        return f"{self.owner} opened a bank account with a starting balance of {self.balance}"

    def __repr__(self):
        return f'{self.owner} is the owner of this account'

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds!")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")
````

````python
bank1 = BankAccount('Alaa', 1000)
````

````python
print(bank1)
````

# 6. Inheritance

So far, every class we've built has stood alone. `Rectangle` doesn't know anything about `BankAccount`, and vice versa. But in the real world — and in real codebases — objects often come in *families*.
<br><br>

Think about the bank account example: a real bank doesn't offer just one kind of account. There are **savings accounts**, **checking accounts**, **student accounts**. All of them are still, fundamentally, bank accounts — they have an owner, a balance, they support deposits and withdrawals. But each kind also has something *extra* or slightly *different*:

- A **savings account** earns interest.
- A **checking account** might allow overdrafts up to a limit.
- A **student account** might waive fees under a certain age.

````python
class Shape: # Shape is called the Parent/Super class
    def __init__(self, color, dimension):
        self.color = color
        self.dimension = dimension

    def change_color(self, color):
        self.color = color

class Square(Shape): # Square is called a Subclass of Shape
    def __init__(self, area, color, dimension):
        super().__init__(color, dimension)

    def change_color(self, color):
        self.color = color
        return f"The color of your square is {self.color}"
    
````

````python
square1 = Square(50, 'red', 2)

square1.color
````

````python
print(square1.change_color('black'))
print(square1.color)
````
