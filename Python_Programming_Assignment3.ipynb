### `Game` Class

### `RockPaperScissors` Class

Since a `RockPaperScissors` game **is a** `Game`, it inherits `n_rounds`, `current_round_num`, `computer_wins`, `user_wins`, `calc_total_winner`, and `play_game` as-is. It adds:

**`__init__(self, n_rounds, current_round_num=1)`**
Calls `super().__init__(n_rounds, current_round_num)`, then sets `self.u_choice = None` and `self.c_choice = None`.

**`computer_choice(self)`**
Randomly returns `"rock"`, `"paper"`, or `"scissors"`.

**`calc_scores_per_round(self, user_choice, computer_choice)`**
Stores `self.u_choice = user_choice` and `self.c_choice = computer_choice`, then, using standard rock-paper-scissors rules, returns a dict `{'user': <score>, 'computer': <score>}` where `1` means that player won the round, `-1` means they lost, and `0` means a tie.

**`calc_winner_per_round(self, scores)`** *(overrides `Game`'s version)*
Prints a message showing what the computer and the user each chose, then calls `super().calc_winner_per_round(scores)` to reuse the parent's score-updating and outcome-announcing logic rather than duplicating it.



````python
class Game:
    def __init__(self, n_rounds: int, current_round_num: int = 1):
        self.n_rounds = n_rounds
        self.current_round_num = current_round_num
        self.computer_wins = 0
        self.user_wins = 0

    def calc_winner_per_round(self, scores: dict):
        self.computer_wins += scores['computer']
        self.user_wins += scores['user']

        if scores['computer'] > scores['user']:
            print("Computer wins this round!!")
        elif scores['computer'] < scores['user']:
            print("You win this round!!")
        else:
            print("It's a tie!!")

    def calc_total_winner(self):
        if self.computer_wins == self.user_wins:
            return "It's a tie"
        elif self.user_wins > self.computer_wins:
            return "You win the game!!"
        else:
            return "Computer wins the game!!"

    def play_game(self, user_choices: list):
        for user_choice in user_choices:
            if self.current_round_num > self.n_rounds:
                break
            computer_choice = self.computer_choice()
            scores = self.calc_scores_per_round(user_choice, computer_choice)
            self.calc_winner_per_round(scores)
            self.current_round_num += 1
        return self.calc_total_winner()


class RockPaperScissors(Game):
    def __init__(self, n_rounds: int, current_round_num: int = 1):
        self.u_choice = ...
        self.c_choice = ...

    def computer_choice(self):
        options = ['rock', 'paper', 'scissors']
        ...

    def calc_scores_per_round(self, user_choice: str, computer_choice: str) -> dict:
        ...

    def calc_winner_per_round(self, scores: dict):
        ...
````

# Question 3 - No Cap, Just `Heap`-ing it Real

Data structures are an important part of both theoretical computer science and software engineering. This question gives you the opportunity to practice `OOP` in the context of data structures.

A heap is a kind of data structure that stores its elements in a way that makes it fast to repeatedly find and remove the smallest (or largest) item, without ever needing to fully sort everything. This is important since many real-world problems only ever need the current best (or worst) item at any given time, not a fully ordered list. Sorting everything up front would waste work you may never use.

Read more about heaps [here](https://www.geeksforgeeks.org/dsa/heap-data-structure/).

### Methods to implement

In this exercise you will implement the following methods:

**`is_empty(self)`**
- Check whether the heap currently has any elements in it.
- Returns `True` if the heap has no elements, `False` otherwise.

**`size(self)`**
- Check how many elements are currently in the heap.
- Returns an `int`.

**`peek(self)`**
- Look at the item at the root of the heap (the smallest item, in a
  min-heap) without removing it.
- Raises `IndexError` if the heap is empty.
- Returns the value at the root.

**`insert(self, value)`**
- Add `value` to the heap.
- Append it to the end of the internal array, then call
  `self._sift_up(...)` to move it into its correct position.
- Returns nothing.

**`_compare(self, a, b)`**
- Decide whether `a` should be considered "better" than `b`, i.e.
  closer to the root, according to `self.is_min_heap`.
- If `self.is_min_heap` is `True`, `a` is better when it is smaller
  than `b`. If `False`, `a` is better when it is larger than `b`.
- Returns a `bool`.

**`heap_sort(self)`**
- Return a **new list** containing every item in the heap, in fully
  sorted order without permanently emptying or damaging the
  original heap (work on a copy).
- Ascending order for a min-heap.
- Returns a `list`.

````python
class Heap:
    def __init__(self, is_min_heap=True):
      self._data = []
      self.is_min_heap = is_min_heap

    def _sift_up(self, i):
      while i > 0:
          parent = (i - 1) // 2
          if self._compare(self._data[i], self._data[parent]):
              self._data[i], self._data[parent] = self._data[parent], self._data[i]
              i = parent
          else:
              break

    def _sift_down(self, i):
      n = len(self._data)
      while True:
          left, right = 2 * i + 1, 2 * i + 2
          best = i
          if left < n and self._compare(self._data[left], self._data[best]):
              best = left
          if right < n and self._compare(self._data[right], self._data[best]):
              best = right
          if best == i:
              break
          self._data[i], self._data[best] = self._data[best], self._data[i]
          i = best

    def extract(self):
      if self.is_empty():
          raise IndexError("extract from empty heap")
      top = self._data[0]
      last = self._data.pop()
      if self._data:
          self._data[0] = last
          self._sift_down(0)
      return top

    def is_empty(self):
      ...

    def size(self):
      ...

    def peek(self):
      ...

    def insert(self, value):
      ...

    def _compare(self, a, b):
      ...

    def heap_sort(self):
      ...
````
