## List comprehensions - building lists in one line

Here's a very common pattern: build a new list by applying an operation to every element of an existing one.

```python
squares = []
for x in range(1, 6):
    squares.append(x ** 2)
```

Python gives us a shorter, more "Pythonic" way to write exactly this : a **list comprehension**:

```python
squares = [x ** 2 for x in range(1, 6)]
```

Same result, one line, and it reads almost like English: *"x squared, for x in range 1 to 6"*.


````python
squares_loop = []
for x in range(1, 6):
    squares_loop.append(x ** 2)
squares_loop
````

````python
squares_comprehension = [x ** 2 for x in range(1, 6)]

print(squares_loop)
print(squares_comprehension)
print(squares_loop == squares_comprehension)
````

You can even add a **condition**:

```python
evens = [x for x in range(1, 21) if x % 2 == 0]
```


````python
evens = [x for x in range(1, 21) if x % 2 == 0]
print(evens)

squared_evens = [x ** 2 for x in evens]
print(squared_evens)
````

### 2.1 Comprehensions work great with dictionaries too

Since dictionary values can be lists, and we can loop over `.items()`, comprehensions combine nicely with what you just learned:


````python
student = {
    "name": "Uwase",
    "age": 20,
    "cohort": 2026,
    "courses": ["Python", "Linear Algebra", "Statistics"]
}

course_lengths = [len(course) for course in student["courses"]]
print(course_lengths)

# names of students in cohort_2026 who love Python
python_lovers = [name for name, info in cohort_2026.items() if info["favorite_subject"] == "Python"]
print(python_lovers)
````

Which of the two lines above feels more readable to you — the `for` loop version or the comprehension? There's no single right answer: comprehensions are great for *short* transformations, but a full `for` loop is often clearer once the logic gets more complex.

## Defining Functions

### 3.0 Built-in vs. user-defined functions

You've already used **built-in functions** without calling them that: `print()`, `len()`, `input()`, `range()`, `int()`, `float()`... Python ships with these ready to use.

Today, you become the author: you'll write **user-defined functions**.

### 3.1 Anatomy of a function

```python
def function_name(parameter1, parameter2, ...):
    """Docstring: what this function does."""
    <block of instructions>
    return result   # optional
```

Let's build one together, live.


Let's turn the multiplication-table-of-8 logic into a function.


````python
def multiplication_table_8():
    """Displays the first 20 values of the multiplication table for 8."""
    n = 1
    while n <= 20:
        v = n * 8
        print(n, 'x', 8, '=', v)
        n += 1
````

Notice: defining the function does **nothing** on its own, nothing printed! To actually run it, we must **call** it:

````python
multiplication_table_8()  # this is the "call" , now it actually runs
````

If I run the cell above **twice in a row**, will it print the table twice, or does Python remember it already printed it once?


### 3.2 Parameters - make it flexible

Hard-coding `8` is limiting. Let's generalize using a **parameter**:


````python
def multiplication_table(base):
    """Displays the multiplication table for <base>, from 1 to 20."""
    n = 1
    while n <= 20:
        v = n * base
        print(n, 'x', base, '=', v)
        n += 1

multiplication_table(2)
print("============================")
multiplication_table(11)
````

````python
multiplication_table(5)
````

### 3.3 Returning a value

So far our functions *print* something but don't give back a usable **value** to the rest of the program. Compare:

```python
def square_print(x):
    print(x ** 2)      # just displays it

def square_return(x):
    return x ** 2       # gives the value back, usable later
```


````python
def square_print(x):
    print(x ** 2)

def square_return(x):
    return x ** 2

a = square_print(4)   # prints 16, but what does `a` contain?
b = square_return(4)  # prints nothing, but `b` contains 16

print("a =", a)
print("b =", b)
````

Why is `a` equal to `None`? What actually happened when Python ran `square_print(4)`?

This is one of the most common beginner bugs: *forgetting `return` and expecting the function's printed value to be usable elsewhere.*


### 3.4 Multiple return values


````python
PI = 3.14159

def surface_volume_sphere(r):
    """Returns (surface_area, volume) of a sphere with radius r."""
    surf = 4.0 * PI * r**2
    vol = surf * r / 3
    return surf, vol
    
    

s, v = surface_volume_sphere(3)
print(f"Surface = {s:.5f}, Volume = {v:.2f}")
````

### 4.2 Passing a function as an argument

Functions are values too, you can pass one function into another!


````python
def my_function(x):
    return 2 * x**3 + x - 5

def tabulate(f, lower, upper, num_steps):
    """Prints f(x) for num_steps evenly spaced values between lower and upper."""
    step = (upper - lower) / num_steps
    x = lower
    while x <= upper:
        print(f"f({x:.2f}) = {f(x):.2f}")
        x += step

tabulate(my_function, 0, 2, 4)
tabulate(abs, -2, 2, 4)   # works with built-ins too!
````

### 4.3 Local vs. Global variables


````python
def my_function_scope():
    p = 20         # local to my_function_scope
    print(p, q)    # q is not defined here... but Python can still SEE the global q

p = 15
q = 38

print(p, q)          # 15 38
my_function_scope()  # 20 38
print(p, q)          # 15 38  <-- p is UNCHANGED outside the function!
````

Why does the *last* `print(p, q)` still show `15`, even though we set `p = 20` inside the function?

**Key idea:** Variables created **inside** a function are *local*: they exist only during that function's call and disappear afterward.

**⚠️ Best practice:** You *can* force a function to modify a global variable with the `global` keyword, but it's usually a code smell — prefer `return`-ing values instead.


### 4.4 Arbitrary arguments: `*args` and `**kwargs`

`*args` collects extra positional arguments into a tuple. `**kwargs` collects extra named arguments into a **dictionary**, which will feel very familiar now!


````python
def total(*args):
    """Returns the sum of any number of numbers."""
    result = 0
    for number in args:
        result += number
    return result

print(total(23))
print(total(23, 42, 13))
````

````python
def describe_student(**kwargs):
    return kwargs

info = describe_student(name="Amina", country="Rwanda", cohort=2026)
print(info)
print(type(info))   # it really is a dict!
````

## Recursion : a function that calls *itself*

A function is allowed to call **itself**. This is called **recursion**, and it's especially natural for sequences defined "in terms of themselves", like Fibonacci!

Recall the mathematical definition from Lesson 0:

$$F_0 = a, \quad F_1 = b, \quad F_{n+2} = F_{n+1} + F_n$$

This is *already* a recursive definition. Let's translate it directly into Python:


````python
def fib_recursive(a, b, n):
    """Returns F_n for the sequence starting with F_0=a, F_1=b, using recursion."""
    if n == 0:
        return a          # base case
    elif n == 1:
        return b          # base case
    else:
        return fib_recursive(a, b, n - 1) + fib_recursive(a, b, n - 2)  # recursive call!

print(fib_recursive(0, 1, 17))  
````

**Every recursive function needs:**
1. A **base case** : where it stops calling itself
2. A **recursive case** : where it calls itself with a *smaller* problem

**Without a base case, a recursive function calls itself forever** : you'll get a `RecursionError`, the recursive cousin of an infinite `while` loop.
